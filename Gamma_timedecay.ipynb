```python
# Python script to simulate Gamma acceleration impact
# For 30 DTE vs 2 DTE near the strike

# Scenario A: 30 Days to Expiration (Smooth Gamma)
stock_start_a = 100.00
delta_start_a = 0.50
gamma_a = 0.04 # Spread out gamma
stock_move = 2.00

# New Delta calculation: Delta_new = Delta_old + (Gamma * Move)
delta_end_a = delta_start_a + (gamma_a * stock_move)
# Estimated price change via Taylor expansion: (Delta * Move) + (0.5 * Gamma * Move^2)
price_change_a = (delta_start_a * stock_move) + (0.5 * gamma_a * (stock_move**2))

# Scenario B: 2 Days to Expiration (Explosive ATM Gamma)
stock_start_b = 100.00
delta_start_b = 0.50
gamma_b = 0.20 # Concentrated gamma close to expiry

delta_end_b = delta_start_b + (gamma_b * stock_move)
price_change_b = (delta_start_b * stock_move) + (0.5 * gamma_b * (stock_move**2))

print(f"30 DTE -> Ending Delta: {delta_end_a:.2f}, Option Value Gain: ${price_change_a:.2f}")
print(f"2 DTE  -> Ending Delta: {delta_end_b:.2f}, Option Value Gain: ${price_change_b:.2f}")
```
